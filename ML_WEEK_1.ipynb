```python
# Cell 1: Import libraries and load dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder

# Load Titanic dataset directly from online URL
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic_df = pd.read_csv(url)
print("Dataset loaded successfully!")
print(titanic_df.head(10))
```

```python
# Cell 2: Assignment Part 1 & 2 - Basic stats and missing values
print("--- INFO ---")
print(titanic_df.info())

print("\n--- DESCRIBE ---")
print(titanic_df.describe())

# Handle missing data using median imputation for Age
titanic_df['Age'] = titanic_df['Age'].fillna(titanic_df['Age'].median())
print("\nMissing values in Age after imputation:", titanic_df['Age'].isnull().sum())
```

```python
# Cell 3: Assignment Part 3 & Mini Project - Encoding and Visualization
# Encode categorical variables
le = LabelEncoder()
titanic_df['Sex'] = le.fit_transform(titanic_df['Sex'])

# One-Hot Encoding for 'Embarked' column using pandas get_dummies
titanic_df = pd.get_dummies(titanic_df, columns=['Embarked'], drop_first=True)

# Visualize age distribution
plt.figure(figsize=(8, 5))
sns.histplot(titanic_df['Age'], kde=True, color='blue')
plt.title('Titanic Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

# Output cleaned dataset as new CSV
titanic_df.to_csv('cleaned_titanic.csv', index=False)
print("Cleaned dataset saved successfully as cleaned_titanic.csv!")
```
